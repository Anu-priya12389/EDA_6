```python
plt.figure(figsize=(10,6))
sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True
)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(
    data=df,
    x="Age",
    hue="Gender",
    bins=15,
    kde=True,
    element="step"
)
plt.title("Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
gender_c=df["Gender"].value_counts()
plt.figure(figsize=(7,5))

sns.barplot(
    x=gender_c.index,
    y=gender_c.values
)
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
condition_c=df["Medical_Condition"].value_counts()
plt.figure(figsize=(7,5))

sns.barplot(
    x=condition_c.values,
    y=condition_c.index
)
plt.title("Medical Condition Distribution")
plt.xlabel("Number of Patients")
plt.ylabel("Medical Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(
    data=df,
    x="Medical_Condition",
    hue="Gender"
)
plt.title("Gender Distribution Across Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.show()
```

```python
admission_counts=df["Admission_Type"].value_counts()
plt.figure(figsize=(8,5))

sns.barplot(
    x=admission_counts.index,
    y=admission_counts.values
)
plt.title("Admission Type Distribution")
plt.xlabel("Admission Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
admission_condition=pd.crosstab(df["Medical_Condition"],df["Admission_Type"])
print(admission_condition)
```

```python
admission_condition.plot(
    kind="bar",
    figsize=(12,6),
    stacked=True
)
plt.title("Admission Type Distribution Across Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.legend(title="Admission_Type")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(
    data=df,
    x="Billing_Amount",
    bins=20,
    kde=True
)
plt.title("Billing Amount Distribution")
plt.xlabel("Billing Amount")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(
    data=df,
    x="Medical_Condition",
    y="Billing_Amount"
)
plt.title("Billing Amount By Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=30)
plt.show()
```

```python
avg_condition_bill=(df.groupby("Medical_Condition")["Billing_Amount"].mean().sort_values(ascending=False))
print(avg_condition_bill)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    x=avg_condition_bill.values,
    y=avg_condition_bill.index)
plt.title("Average Billing Amount by Medical Condition")
plt.xlabel("Average Billing Amount")
plt.ylabel("Medical Condition")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.violinplot(
    data=df,
    x="Insurance_Provider",
    y="Billing_Amount"
)
plt.title("Billing Amount Distribution by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=30)
plt.show()
```

```python
insurance_billing=(
    df.groupby("Insurance_Provider")["Billing_Amount"]
    .mean()
    .sort_values(ascending=False)
)
print(insurance_billing)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    x=insurance_billing.values,
    y=insurance_billing.index
)
plt.title("Average Billing Amount by Insurance Provider")
plt.xlabel("Average Billing Amount")
plt.ylabel  ("Insurance Provider")
plt.show()
```

```python
df["Admission_Date"]=pd.to_datetime(df["Admission_Date"])
df["Admission_Month"]=(df["Admission_Date"].dt.to_period("M")
)
```

```python
print(df["Admission_Month"].head())
```

```python
monthly_admissions=df.groupby("Admission_Month").size().reset_index(name="Admissions")
```

```python
monthly_admissions["Admission_Month"]=(monthly_admissions["Admission_Month"].dt.to_timestamp())
```

```python
print(monthly_admissions)
```

```python
plt.figure(figsize=(14,6))
plt.plot(
    monthly_admissions["Admission_Month"],
    monthly_admissions["Admissions"],
    marker="o"
)
plt.title("Monthly Admissions Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
